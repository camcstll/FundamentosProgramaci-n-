```python

# ========================================
# SISTEMA DE PRESTAMOS
# Version 3 - Historial de prestamos y necesidades de software
# ========================================

from datetime import datetime

# Lista de alumnos: numero de control, nombre y carrera
alumnos = [ "26040200", "26040201", "26040202", "26040203", "26040204"]


# Cada computadora guarda: ID, aula, software y estado
computadoras = [
    ["PC-LC1-01", "LC1", "Python, VS Code", "Disponible"],
    ["PC-LC1-02", "LC1", "Python, Java, VS Code", "Prestada"],
    ["PC-LC1-03", "LC1", "AutoCAD, MATLAB", "Reservada"],
    ["PC-LC2-01", "LC2", "Python, VS Code", "Disponible"],
    ["PC-LC2-02", "LC2", "Python, Java, VS Code", "Prestada"],
    ["PC-LC2-03", "LC2", "AutoCAD, MATLAB", "Disponible"],
    ["PC-LC3-01", "LC3", "Python, VS Code", "Disponible"],
    ["PC-LC3-02", "LC3", "Python, Java, VS Code", "Prestada"],
    ["PC-LC3-03", "LC3", "AutoCAD, MATLAB", "Reservada"],
    ["PC-LC3-04", "LC3", "Python, VS Code", "Mantenimiento"]]

# Este es el historial de prestamos
prestamos = []

# Para marcar el istorial de necesidades de software
necesidades_software = []


def listar_computadoras():
    print("\n--- COMPUTADORAS ---")

    for i in range(len(computadoras)):
        print(i + 1, "Aula:", computadoras[i][1],
              "| ID:", computadoras[i][0],
              "| Software:", computadoras[i][2],
              "| Estado:", computadoras[i][3])


def buscar_por_software():
    programa = input("\nIngrese el software que desea buscar: ").strip()

    print("\n--- RESULTADOS ---")
    encontrado = False

    for i in range(len(computadoras)):
        if programa.lower() in computadoras[i][2].lower():
            print(i + 1, "ID:", computadoras[i][0],
                  "| Aula:", computadoras[i][1],
                  "| Software:", computadoras[i][2],
                  "| Estado:", computadoras[i][3])
            encontrado = True

    if encontrado == False:
        print("No se encontraron computadoras con ese software.")


def solicitar_prestamo():
    print("\n=== SOLICITAR PRESTAMO ===")
    num_control = input("Numero de control del alumno: ").strip()
    alumno_encontrado = False
    # Buscar el numero de control en la primera columna
    for i in range(len(alumnos)):
        if num_control == alumnos[i][0]:
            alumno_encontrado = True
            nombre_alumno = alumnos[i][1]
            break

    if alumno_encontrado:
        print("Alumno:", nombre_alumno)
        programa = input("Que programa necesita: ").strip()
        if programa == "":
            print("Debe escribir el nombre del programa.")
            return

        encontrado = False
        for i in range(len(computadoras)):
            programas_instalados = computadoras[i][2].lower()
            lista_programas = [
                p.strip().lower()
                for p in computadoras[i][2].split(",")
            ]

            if programa.lower() in lista_programas:
                encontrado = True
                if computadoras[i][3] == "Disponible":
                    computadoras[i][3] = "Prestada"
                    nuevo_prestamo = [
                        num_control,
                        nombre_alumno,
                        programa,
                        computadoras[i][1],
                        computadoras[i][0]
                    ]

                    prestamos.append(nuevo_prestamo)

                    print("Se asigno la computadora",
                          computadoras[i][0],
                          "que esta en el aula",
                          computadoras[i][1])

                    print("El prestamo se guardo en el historial.")
                    return

        if encontrado:
            print("El software existe, pero no hay computadoras disponibles.")
        else:
            print("El programa no se encontro en las computadoras.")
            respuesta = input("Desea agregarlo a la lista de necesidades de software? (si/no): "
            ).strip().lower()

            if respuesta == "si" or respuesta == "sí":
                fecha_hora = datetime.now().strftime("%d-%m-%Y %H:%M:%S")

                nueva_necesidad = [programa, fecha_hora]
                necesidades_software.append(nueva_necesidad)

                print("Necesidad de software registrada correctamente.")
                print("Programa:", programa)
                print("Fecha:", fecha_hora)
            else:
                print("No se agrego el programa a la lista.")

    else:
        print("El alumno no existe en la lista.")


def devolver_computadora():
  print("\n ===DEVOLVER COMPUTADORA====")
  id_pc = input("Dame el ID de la computadora")
  existe = False
  for i in range( len(computadoras) ):
    if computadoras[i][0] == id_pc:
      existe = True
      if computadoras[i][3]== "Prestada" or computadoras[i][3]=="Reservada":
        computadoras[i][3] = "Disponible"
        print("La computadora", id_pc, "ya esta disponible")
        break
      else:
        print("La computadora", id_pc, "no esta prestada/reservada")
  if existe == False:
    print("No existe la computadora", id_pc)


def crear_reservacion():
    print("\n=== CREAR RESERVACION ===")

    num_control = input("Numero de control del alumno: ").strip()
    alumno_encontrado = False

    for i in range(len(alumnos)):
        if num_control == alumnos[i][0]:
            alumno_encontrado = True
            nombre_alumno = alumnos[i][1]
            break

    if alumno_encontrado:
        listar_computadoras()

        id_pc = input("\nID de la computadora que desea reservar: ").strip()

        for i in range(len(computadoras)):
            if computadoras[i][0] == id_pc:
                if computadoras[i][3] == "Disponible":
                    computadoras[i][3] = "Reservada"
                    print("Computadora reservada correctamente para",
                          nombre_alumno)
                else:
                    print("La computadora no esta disponible.")
                return

        print("No existe esa computadora.")

    else:
        print("El alumno no existe en la lista.")


def agregar_computadora():
  print("\n--- AGREGAR COMPUTADORA ---")
  id_computadora = input("ID de la computadora: ")
  aula = input("Aula: ")
  programas = input("Software instalado: ")
  nueva_computadora = [id_computadora, aula, programas, "Disponible"]
  computadoras.append(nueva_computadora)
  print("\nComputadora agregada correctamente.")


def historial_prestamos():
    print("\n--- HISTORIAL DE PRESTAMOS DE COMPUTADORAS ---")

    if len(prestamos) == 0:
        print("Todavia no hay prestamos registrados.")
    else:
        for i in range(len(prestamos)):
            print(
                i + 1,
                "| Numero de control:", prestamos[i][0],
                "| Alumno:", prestamos[i][1],
                "| Software:", prestamos[i][2],
                "| Aula:", prestamos[i][3],
                "| ID:", prestamos[i][4])


def historial_necesidades():
    print("\n--- NECESIDADES DE SOFTWARE ---")
    for i in range(len(necesidades_software)):
            print(
                i + 1,
                "| Programa:", necesidades_software[i][0],
                "| Fecha:", datetime, necesidades_software[i][1])

    else:
        print("Todavia no hay necesidades de software registradas ahorita")


# ========================================
# MENU PRINCIPAL
# ========================================

opcion = 0
while opcion <=8:
    print("\n================================")
    print("      SISTEMA DE PRESTAMOS")
    print("================================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar prestamo")
    print("4. Devolver computadora")
    print("5. Crear reservacion")
    print("6. Agregar computadora")
    print("7. Historial de prestamos")
    print("8. Necesidades de software")
    print("9. Salir")
    print("================================")
    opcion = int(input("Seleccione una opcion: "))


    if opcion == 1:
        listar_computadoras()

    elif opcion == 2:
        buscar_por_software()

    elif opcion == 3:
        solicitar_prestamo()

    elif opcion == 4:
        devolver_computadora()

    elif opcion == 5:
        crear_reservacion()

    elif opcion == 6:
        agregar_computadora()

    elif opcion == 7:
        historial_prestamos()

    elif opcion == 8:
        historial_necesidades()

    elif opcion == 9:
        print("\nGracias por utilizar el sistema.")

    else:
        print("Opcion no valida.")
```
